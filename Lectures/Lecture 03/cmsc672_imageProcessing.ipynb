```python
import numpy as np
import imageio.v2 as io
import scipy
import cv2
import matplotlib
import matplotlib.pyplot as plt
```

```python
image_url = "https://mountvernonplace.org/wp-content/uploads/2021/07/monument-1.jpg"

# read image
x = io.imread(image_url, pilmode="RGB").astype(np.uint8)

# display image
plt.imshow(x)

# show image dimensions
# print(x.shape)

# print(np.max(x), np.min(x))
```

```python
# pointwise operations
## lighten, darken, invert
x_darken = np.clip(x - [128, 128, 128], 0, 255)
x_lighten = np.clip(x + [128, 128, 128], 0, 255)
x_invert = 255 - x
x_contrast_up = np.clip(x * 1.5, 0, 255).astype(np.uint8)
x_contrast_down = np.clip(x /2, 0, 255).astype(np.uint8)

fig, axs = plt.subplots(1, 6, figsize=(16, 16))
for ax in axs:
    ax.set_axis_off()
axs[0].set_title("Original")
axs[1].set_title("Darken")
axs[2].set_title("Lighten")
axs[3].set_title("Invert")
axs[4].set_title("Contrast Up")
axs[5].set_title("Contrast Down")

axs[0].imshow(x)
axs[1].imshow(x_darken)
axs[2].imshow(x_lighten)
axs[3].imshow(x_invert)
axs[4].imshow(x_contrast_up)
axs[5].imshow(x_contrast_down)
plt.show()
```

```python
# @title
# question from Garvita after class: can we do gradual lightening/darkening from top to bottom

# setp 1: secret sauce (Tejas' numpy mastery to produce a gradually increasing matrix of the same size as the image)
a = np.linspace((0, 0, 0), (255, 255, 255), num=x.shape[0])
a = np.expand_dims(a, axis=1)
a = np.tile(a, (1, x.shape[1], 1))

# we will now add or subtract this matrix "a" to lighten or darken the image gradually from top to bottom
x_gradually_darken = np.clip(x - a, 0, 255).astype(np.uint8)
x_gradually_lighten = np.clip(x + a, 0, 255).astype(np.uint8)
fig, axs = plt.subplots(1, 3, figsize=(16, 16))
for ax in axs:
    ax.set_axis_off()
axs[0].set_title("Original")
axs[1].set_title("Gradually Darken")
axs[2].set_title("Gradually Lighten")

axs[0].imshow(x)
axs[1].imshow(x_gradually_darken)
axs[2].imshow(x_gradually_lighten)
plt.show()

# question:  is this really "pointwise processing"?
```

```python
# Gaussian filter
x_g_1 = cv2.GaussianBlur(x, (41, 41), 1)
x_g_2 = cv2.GaussianBlur(x, (41, 41), 5)
x_g_3 = cv2.GaussianBlur(x, (41, 41), 10)
x_g_4 = cv2.GaussianBlur(x, (41, 41), 20)

# Laplacian Filter
x_l_1 = x - x_g_1
x_l_2 = x - x_g_2
x_l_3 = x - x_g_3
x_l_4 = x - x_g_4



fig, axs = plt.subplots(2, 5, figsize=(12, 4))
for ax in axs.flatten():
    ax.set_axis_off()
axs[0, 0].set_title("Original")
axs[0, 1].set_title("Sigma = 1")
axs[0, 2].set_title("Sigma = 5")
axs[0, 3].set_title("Sigma = 10")
axs[0, 4].set_title("Sigma = 20")

axs[0, 0].imshow(x)
axs[0, 1].imshow(x_g_1)
axs[0, 2].imshow(x_g_2)
axs[0, 3].imshow(x_g_3)
axs[0, 4].imshow(x_g_4)

axs[1, 1].imshow(x_l_1)
axs[1, 2].imshow(x_l_2)
axs[1, 3].imshow(x_l_3)
axs[1, 4].imshow(x_l_4)
plt.show()
```
